```python
#Inicio del juego

def start_game():

    print("Estás visitando museo del prado y te pierdes de tus amigos. ¡Te están enviando mensajes y dicen que te están esperando al salir!")

start_game()
```

```python
# Funciones

def buscar_en_sala(nombre_item, sala):
    """Busca un ítem por su nombre dentro de una sala. Devuelve el ítem o None."""
    for item in object_relations[sala["name"]]:
        if item["name"].lower() == nombre_item.lower():
            return item
    return None

def puede_abrir(puerta, game_state):
    """True si alguna llave del jugador tiene esa puerta como target."""
    for llave in game_state["keys_collected"]:
        if llave["target"] == puerta:
            return True
    return False
 
 
def cruzar_puerta(puerta, game_state):
    """Intenta cruzar una puerta. Si el jugador tiene la llave, devuelve la sala
    del otro lado. Si no, avisa de que está cerrada y devuelve None."""
    if not puede_abrir(puerta, game_state):
        print(f"La {puerta['name']} está cerrada. Necesitas su llave.")
        return None
 
    sala_a, sala_b = object_relations[puerta["name"]]   # las dos salas que une
    if game_state["current_room"] == sala_a:
        return sala_b
    return sala_a
```
