```python
# Funciones

def buscar_en_sala(nombre_item, sala):
    """Busca un ítem por su nombre dentro de una sala. Devuelve el ítem o None."""
    for item in object_relations[sala["name"]]:
        if item["name"].lower() == nombre_item.lower():
            return item
    return None

def puede_abrir(puerta, game_state):
    """True si alguna llave del jugador tiene esa puerta como target."""
    for llave in game_state["keys_collected"]:
        if llave["target"] == puerta:
            return True
    return False
 
 
def cruzar_puerta(puerta, game_state):
    """Intenta cruzar una puerta. Si el jugador tiene la llave, devuelve la sala
    del otro lado. Si no, avisa de que está cerrada y devuelve None."""
    if not puede_abrir(puerta, game_state):
        print(f"La {puerta['name']} está cerrada. Necesitas su llave.")
        return None
 
    sala_a, sala_b = object_relations[puerta["name"]]   # las dos salas que une
    if game_state["current_room"] == sala_a:
        return sala_b
    return sala_a
```

```python
# ------------------------------------------------------------
# SALA 1: VELÁZQUEZ (y la sala trampa: el aseo)
# ------------------------------------------------------------
def examinar_item_velazquez(nombre_item, game_state):
    """Examina un ítem de la sala Velázquez.
    Devuelve la sala a la que va el jugador si cruza una puerta, o None si se queda."""
    item = buscar_en_sala(nombre_item, sala_velazquez)
 
    if item is None:
        print("Aquí no hay nada con ese nombre.")
        return None
 
    if item["type"] == "door":
        return cruzar_puerta(item, game_state)
 
    coger_contenido(item, game_state)
    return None
 
 
def visitar_aseo():
    """Sala trampa: el jugador entra, no hay nada y tiene que volver."""
    print("\nAbres la puerta A... es un aseo. Solo hay un lavabo y un espejo.")
    print("Aquí no hay salida. Vuelves a la sala Velázquez.")
 
 
def jugar_velazquez(game_state):
    """El jugador explora la sala Velázquez hasta que cruza la puerta B."""
    game_state["current_room"] = sala_velazquez
    print("\nEstás en la sala Velázquez del Museo del Prado. Las puertas se han cerrado tras de ti.")
 
    while True:
        nombres = [item["name"] for item in object_relations["sala_velazquez"]]
        print("Ves: " + ", ".join(nombres))
        eleccion = input("¿Qué quieres examinar? ").strip()
        siguiente_sala = examinar_item_velazquez(eleccion, game_state)
 
        if siguiente_sala == aseo:
            visitar_aseo()                      # entra, ve que no hay nada y vuelve
        elif siguiente_sala == sala_musas:
            game_state["current_room"] = sala_musas
            print("Abres la puerta B y pasas a la Sala de las Musas.")
            return
```

```python
# FUNCIONES DEL JUEGO (Sala 2):

import random

# Función para imprimir un salto de línea
def linebreak():
    print("\n")

# Función para iniciar el juego
def start_game():
    # Bienvenida al juego
    print("Has cruzado la primera puerta del Museo del Prado, ahora te encuentras en la Sala de las Musas.")
    play_room(INIT_GAME_STATE["current_room"])

# Función para jugar en una sala
def play_room(room):
    # Actualizar la sala actual en el estado del juego
    INIT_GAME_STATE["current_room"] = room
    # Comprobar si el jugador ha llegado a la sala objetivo
    if INIT_GAME_STATE["current_room"] == INIT_GAME_STATE["target_room"]:
        print("¡Felicidades! Has llegado a la Cafetería del Museo del Prado.")
        return

# Función para jugar en una sala
    print(f"Estás en la {room['name']}.")
    # Preguntar al jugador qué quiere hacer
    accion = input("¿Qué quieres hacer? Escribe 'explorar' o 'examinar': ").strip().lower()
    # Comprobar la acción del jugador
    if accion == "explorar":
        explore_room(room)
        linebreak()
        play_room(room)
    elif accion == "examinar":
        examine_item(input("¿Qué quieres examinar?: ").strip().lower())
    else:
        print("No te entiendo. Escribe 'explorar' o 'examinar'.")
        linebreak()
        play_room(room)

# Función para explorar una sala
def explore_room(room):
    # Mostrar los objetos en la sala
    items = [i["name"] for i in object_relations[room["name"]]]
    print(f"Exploras la sala. Ves: {', '.join(items)}.")

# Función para obtener la siguiente sala a través de una puerta
def get_next_room_of_door(door, current_room):
    # Buscar la sala que no es la actual en las relaciones del objeto puerta
    for room in object_relations[door["name"]]:
        if room != current_room:
            return room

# Función para adivinar un número
def adivinar_numero():
    # Solicitar al jugador que adivine un número del 1 al 5
    while True:
        respuesta = input("Adivina un número del 1 al 5: ")
        # Comprobar que es un número
        try:
            numero = int(respuesta)
        except ValueError:
            print("Error: eso no es un número. Inténtalo otra vez.")
            continue

        # Comprobar que está entre 1 y 5
        if numero < 1 or numero > 5:
            print("Error: el número tiene que estar entre 1 y 5. Inténtalo otra vez.")
            continue
        # Si llega aquí, el número es válido y salimos del bucle
        break

    # Comprobar si ha acertado
    if numero == carlos["secret_number"]:
        print("¡Correcto! Pista: " + pista_carlos["text"])
    else:
        print("Número incorrecto. Puedes volver a intentarlo examinándolo de nuevo.")

# Función para adivinar si un número es par o impar
def par_o_impar():
    numero = random.randint(1, 100)
    print("El vigilante te dice: 'Te doy una pista si aciertas esto'.")

    # Error handling: solo acepta 'par' o 'impar'
    while True:
        respuesta = input("¿El número " + str(numero) + " es par o impar?: ").lower()
        if respuesta == "par" or respuesta == "impar":
            break
        print("Error: escribe 'par' o 'impar'. Inténtalo otra vez.")

    # Calcular la solución
    if numero % 2 == 0:
        solucion = "par"
    else:
        solucion = "impar"

    # Comprobar si ha acertado
    if respuesta == solucion:
        print("¡Correcto! Pista: " + pista_vigilante["text"])
    else:
        print("Incorrecto. Habla otra vez con el vigilante para volver a intentarlo.")

# Función para examinar un objeto
def examine_item(item_name):
    current_room = INIT_GAME_STATE["current_room"]
    next_room = None
    encontrado = False

    # Buscar el objeto en la sala actual
    for item in object_relations[current_room["name"]]:
        if item["name"] != item_name:
            continue
        encontrado = True
        print(f"Examinas {item_name}.")

        # Si es una puerta, comprobar si tienes la llave
        if item["type"] == "door":
            tiene_llave = any(k["target"] == item for k in INIT_GAME_STATE["keys_collected"])
            if tiene_llave:
                print("Tienes la llave. La puerta se abre.")
                next_room = get_next_room_of_door(item, current_room)
            else:
                print("Está cerrada y no tienes la llave.")

        # Carlos V: acertijo del número
        elif item.get("pista_carlos"):
            print(item["descripcion"])
            adivinar_numero()

        # Vigilante: acertijo de par o impar
        elif item.get("pista_vigilante"):
            print(item["descripcion"])
            par_o_impar()

        # Si es un mueble normal, mostrar la descripción y comprobar si tiene llaves
        else:
            print(item["descripcion"])
            if object_relations[item["name"]]:
                llave = object_relations[item["name"]].pop()
                INIT_GAME_STATE["keys_collected"].append(llave)
                print(f"¡Encuentras la {llave['name']}!")
            else:
                print("No hay nada más.")
        break

    # Si no se encontró el objeto, informar al jugador
    if not encontrado:
        print("Ese objeto no está en esta sala.")

    # Preguntar al jugador si quiere pasar a la siguiente sala si hay una puerta abierta
    linebreak()
    if next_room and input(f"¿Quieres pasar a la {next_room['name']}? 'si' o 'no': ").strip().lower() == "si":
        play_room(next_room)
    else:
        play_room(current_room)

# Función principal del juego
if __name__ == "__main__":
    start_game()
```

```python
# ------------------------------------------------------------
# SALA 4: VESTUARIO DE EMPLEADOS
# ------------------------------------------------------------
def pedir_codigo(item):
    """Enseña la pista y da 3 intentos para el código. True si acierta, False si no."""
    print(item["hint"])
    print(f"La {item['name']} tiene un candado de números. Tienes 3 intentos.")
 
    for numero_intento in range(1, 4):   # 1, 2 y 3
        intento = input(f"Intento {numero_intento} de 3. Escribe el código: ").strip()
        if intento == item["code"]:
            print("¡Clic! El candado se abre.")
            return True
        print("Código incorrecto.")
 
    print("Has gastado los 3 intentos. Vuelve a intentarlo más tarde.")
    return False
 
 
def examinar_item_vestuario(nombre_item, game_state):
    """Examina un ítem del vestuario. Devuelve True si el jugador elige salir por la puerta D."""
    item = buscar_en_sala(nombre_item, vestuario)
 
    if item is None:
        print("Aquí no hay nada con ese nombre.")
        return False
 
    if item["type"] == "door":
        print("Es la puerta D. Por aquí vuelves a la cafetería.")
        return True
 
    # Si el ítem tiene candado, hay que acertar el código antes de mirar dentro
    if "code" in item:
        if not pedir_codigo(item):
            return False
 
    coger_contenido(item, game_state)
    return False
 
 
def jugar_vestuario(game_state):
    """El jugador explora el vestuario hasta que vuelve a la cafetería por la puerta D."""
    game_state["current_room"] = vestuario
    print("\nEntras en el vestuario de empleados. Huele a café y a sudor.")
 
    salir = False
    while not salir:
        nombres = [item["name"] for item in object_relations["vestuario de empleados"]]
        print("Ves: " + ", ".join(nombres))
        eleccion = input("¿Qué quieres examinar? ").strip()
        salir = examinar_item_vestuario(eleccion, game_state)
 
    game_state["current_room"] = cafeteria
    print("Vuelves a la cafetería.")
```
